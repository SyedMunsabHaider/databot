```python
from databot import DataBot

bot = DataBot("sales_data.csv")
bot.say("Which product made us the most money?")
bot.say("revenue of Laptop in Karachi in 2025")
print(bot.help_text())
```

```python
hr = DataBot("hr_employees.csv")
hr.say("average salary by department")
hr.say("who has the highest salary?")
hr.say("is there a relationship between age and salary?")
```

```python
students = DataBot("student_results.csv")
students.say("how many students have gpa above 3.5?")
students.say("does attendance affect gpa?")
```

### Chat mode
Type a question and press Enter. Type `exit` to stop.

```python
while True:
    q = input("You > ")
    if q.lower() in ("exit", "quit"):
        break
    bot.say(q)
```
